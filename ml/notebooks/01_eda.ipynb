# 01 · Análisis exploratorio del manifest

Requiere el manifest generado con `python -m potato_blight.prepare --config configs/base.yaml`, que solo indexa
datasets con la **licencia verificada** en `datasets.yaml`. Etiquetas por síntoma (D-01).

```python
import json
from pathlib import Path

import pandas as pd

manifest = Path("../data/manifest.csv")
df = pd.read_csv(manifest)
informe = json.loads(manifest.with_suffix(".informe.json").read_text(encoding="utf-8"))
informe
```

```python
# Distribución de clases por dataset (las clases suelen estar desbalanceadas: el entrenamiento usa pesos por clase)
pd.crosstab(df["dataset_id"], df["label"], margins=True)
```

```python
# Imágenes por partición y clase (la validación cruzada está agrupada: un grupo cae en una sola partición)
pd.crosstab(df[df["fold"] >= 0]["fold"], df[df["fold"] >= 0]["label"])
```

```python
# Grupos con varias imágenes (casi duplicados o misma planta) e imágenes compartidas entre datasets
tam = df.groupby("group_id").size()
print("Grupos con más de una imagen:", int((tam > 1).sum()), "· imágenes compartidas entre datasets:", int(df["cross_dataset"].sum()))
df[df["cross_dataset"]].head(20)
```
